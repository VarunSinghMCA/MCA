```python
import pandas as pd

# Load your dataset
dataset = pd.read_csv("D:/Sx Varun/Sx Code/MCA/2nd_Tri/python/dataSet/employee_compensation.csv")
dataset1 = dataset

# Clean numeric columns – remove commas and convert to float
cols_to_clean = ['REGULAR', 'RETRO', 'OTHER', 'OVERTIME', 'TOTAL GROSS']
for col in cols_to_clean:
    dataset1[col] = (dataset1[col]
                     .astype(str)
                     .str.replace(",", "", regex=False)
                     .replace("", 0)
                     .astype(float))

dataset1.head()
```

```python
# 1 Rows and Columns
dataset1.shape
```

```python
# 2 EMP1 – Employee Name, Department, Total Earnings
EMP1 = dataset1[['NAME', 'DEPARTMENT_NAME', 'TOTAL GROSS']]
EMP1.head()

```

```python
# 3 EMP2 – all variables except Overtime
EMP2 = dataset1.drop(columns=['OVERTIME'])
EMP2.head()

```

```python
# 4 Employees with Total >100K and Overtime >5K
high_earners = dataset1[(dataset1['TOTAL GROSS'] > 100000) & (dataset1['OVERTIME'] > 5000)]
len(high_earners)
```

```python
# 5 EMP3 – only Finance department employees
EMP3 = dataset1[dataset1['DEPARTMENT_NAME'] == 'Finance']
EMP3.head()
```

```python
# 6 EMP4 – sort descending by Total Earnings
EMP4 = dataset1.sort_values(by='TOTAL GROSS', ascending=False)
EMP4.head(10)
```

```python
# 7 Add Annual_Bonus column = 10% of Regular Earnings
EMP2['Annual_Bonus'] = EMP2['REGULAR'] * 0.10
EMP2.head()
```

```python
# 8 Average Regular & Overtime – entire dataset and per department
# 8(i) Entire dataset
avg_entire = dataset1[['REGULAR', 'OVERTIME']].mean()
avg_entire

# 8(ii) By department
avg_by_dept = dataset1.groupby('DEPARTMENT_NAME')[['REGULAR', 'OVERTIME']].mean()
avg_by_dept

```

```python
# 9 Top 5 highest-paid employees
top5 = dataset1.nlargest(5, 'TOTAL GROSS')[['NAME', 'TOTAL GROSS']]
top5
```

```python
# 10 Lowest Total Earnings per department
lowest_per_dept = dataset1.loc[dataset1.groupby('DEPARTMENT_NAME')['TOTAL GROSS'].idxmin()][['DEPARTMENT_NAME', 'NAME', 'TOTAL GROSS']]
lowest_per_dept
```

```python
# 11 Department with highest average Overtime
dept_highest_overtime = (dataset1.groupby('DEPARTMENT_NAME')['OVERTIME']
                         .mean()
                         .idxmax())
dept_highest_overtime
```

```python
# 12 Add Level column (Low/Medium/High) in EMP2
def level_category(x):
    if x < 50000:
        return 'Low'
    elif x < 100000:
        return 'Medium'
    else:
        return 'High'

EMP2['Level'] = dataset1['TOTAL GROSS'].apply(level_category)
EMP2.head()
```
