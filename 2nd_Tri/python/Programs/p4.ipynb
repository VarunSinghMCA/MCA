```python
import random
import matplotlib.pyplot as plt
```

```python
# Number of trials
N = 10000

# Counters
sum_7_count = 0
both_heads_count = 0

# Simulate two dice rolls and coin tosses
for _ in range(N):
    # Dice roll simulation (1–6)
    die1 = random.randint(1, 6)
    die2 = random.randint(1, 6)

    # Coin toss simulation ('H' or 'T')
    coin1 = random.choice(['H', 'T'])
    coin2 = random.choice(['H', 'T'])

    # Check for sum = 7
    if die1 + die2 == 7:
        sum_7_count += 1

    # Check if both coins are heads
    if coin1 == 'H' and coin2 == 'H':
        both_heads_count += 1

# Compute probabilities
prob_sum_7 = sum_7_count / N
prob_both_heads = both_heads_count / N

print(f"Simulated Probability of sum = 7: {prob_sum_7:.4f}")
print(f"Theoretical Probability of sum = 7: 6/36 = {6/36:.4f}")
print(f"Simulated Probability of both heads: {prob_both_heads:.4f}")
print(f"Theoretical Probability of both heads: 1/4 = {1/4:.4f}")

# Plot histogram for dice sums
sums = [random.randint(1,6) + random.randint(1,6) for _ in range(N)]
plt.hist(sums, bins=range(2,14), edgecolor='black', align='left')
plt.title("Frequency Distribution of Sum of Two Dice")
plt.xlabel("Sum")
plt.ylabel("Frequency")
plt.show()
```

```python
# Number of trials
N = 10000

# Counters for events
sum_7_count = 0
both_heads_count = 0

# Simulation loop
for _ in range(N):
    # Simulate two dice rolls
    die1 = random.randint(1, 6)
    die2 = random.randint(1, 6)
    total = die1 + die2

    # Simulate two coin tosses
    coin1 = random.choice(['H', 'T'])
    coin2 = random.choice(['H', 'T'])

    # Event 1: sum of dice = 7
    if total == 7:
        sum_7_count += 1

    # Event 2: both coins show heads
    if coin1 == 'H' and coin2 == 'H':
        both_heads_count += 1

# Calculate simulated probabilities
prob_sum_7 = sum_7_count / N
prob_both_heads = both_heads_count / N

# Print results
print("Simulation Results:")
print(f"Total trials: {N}")
print(f"Simulated Probability (sum = 7)         : {prob_sum_7:.4f}")
print(f"Theoretical Probability (sum = 7)       : {6/36:.4f}")
print(f"Simulated Probability (both heads)      : {prob_both_heads:.4f}")
print(f"Theoretical Probability (both heads)    : {1/4:.4f}")

# Optional: visualize frequency of dice sums
sums = [random.randint(1, 6) + random.randint(1, 6) for _ in range(N)]

plt.hist(sums, bins=range(2, 14), edgecolor='black', align='left')
plt.title("Frequency Distribution of Sum of Two Dice")
plt.xlabel("Sum of Dice")
plt.ylabel("Frequency")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
