```python
# First DataFrame
df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Score': [85, 90, 78, 92]
})

# Second DataFrame
df2 = pd.DataFrame({
    'ID': [3, 4, 5, 6],
    'Subject': ['Math', 'Science', 'History', 'English'],
    'Grade': ['B', 'A', 'C', 'B']
})

print("DataFrame 1:")
print(df1)
print("\nDataFrame 2:")
print(df2)
```

```python
inner_join = pd.merge(df1, df2, on='ID', how='inner')
print("\nInner Join:")
print(inner_join)
```

```python
left_join = pd.merge(df1, df2, on='ID', how='left')
print("\nLeft Join:")
print(left_join)
```

```python
right_join = pd.merge(df1, df2, on='ID', how='right')
print("\nRight Join:")
print(right_join)
```

```python
outer_join = pd.merge(df1, df2, on='ID', how='outer')
print("\nOuter Join:")
print(outer_join)
```

```python
import pandas as pd

# First DataFrame
df1 = pd.DataFrame({
    'id': [1, 2, 3, 4],
    'dept': ['IT', 'HR', 'Finance', 'IT'],
    'salary': [50000, 60000, 55000, 58000]
})

# Second DataFrame
df2 = pd.DataFrame({
    'id': [1, 2, 3, 4, 5],
    'dept': ['IT', 'HR', 'Finance', 'Marketing', 'IT'],
    'bonus': [5000, 4000, 6000, 3000, 7000]
})

# Inner join on two keys: id and dept
merged_df = pd.merge(df1, df2, on=['id', 'dept'], how='inner')

print(merged_df)
```
