```python
#dataset.plot(kind='line', x='SepalLength', y=['SepalWidth', 'PetalLength', 'PetalWidth'])

# Scatter plot of SepalLength vs SepalWidth
# dataset.plot(kind='scatter', x='SepalLength', y='SepalWidth', color='red', label='SepalWidth')

# Additional line plots for PetalLength and PetalWidth using matplotlib
# plt.plot(dataset['SepalLength'], dataset['PetalLength'], color='blue', label='PetalLength')
# plt.plot(dataset['SepalLength'], dataset['PetalWidth'], color='green', label='PetalWidth')

plt.scatter(dataset['SepalLength'], dataset['SepalWidth'], color='red', label='SepalWidth')

# Chart formatting
plt.title('Iris Dataset Features vs SepalLength')
plt.xlabel('SepalLength')
plt.ylabel('Value')
plt.grid()
plt.legend()
plt.show()
```

```python
# 1️⃣ Scatter plot for SepalLength vs SepalWidth (red dots)
plt.figure(figsize=(10, 6))
plt.scatter(dataset['SepalLength'], dataset['SepalWidth'],
            color='red', label='SepalWidth (scatter)')

# 2️⃣ Line plot for SepalLength vs PetalLength (blue line)
plt.plot(dataset['SepalLength'], dataset['PetalLength'],
         color='blue', label='PetalLength (line)', linestyle='--', marker='o')

# 3️⃣ Line plot for SepalLength vs PetalWidth (green line)
plt.plot(dataset['SepalLength'], dataset['PetalWidth'],
         color='green', label='PetalWidth (line)', linestyle=':', marker='x')

# Chart formatting
plt.title('Iris Dataset Features vs SepalLength')
plt.xlabel('SepalLength')
plt.ylabel('Feature Values')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python

plt.scatter(dataset['SepalLength'], dataset['SepalWidth'], color='red', label='SepalWidth')
plt.scatter(dataset['SepalLength'], dataset['PetalLength'], color='blue', label='PetalLength')

# Chart formatting
plt.title('Iris Dataset Features vs SepalLength')
plt.xlabel('SepalLength')
plt.ylabel('Value')
plt.grid()
plt.legend()
plt.show()
```

```python
mean_sepal_length = dataset.groupby('Species')['SepalLength'].mean()

colors = ['skyblue', 'lightgreen', 'salmon']

mean_sepal_length.plot(kind='bar', color=colors)

# Chart formatting
plt.title('Mean Sepal Length by Species')
plt.xlabel('Species')
plt.ylabel('Mean Sepal Length')
plt.grid(axis='y', linestyle='--', alpha=0.7)

# for index, value in enumerate(mean_sepal_length):
    # plt.text(index, value + 0.02, round(value, 2), ha='center', fontsize=9)

plt.show()
```
