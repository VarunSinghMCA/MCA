```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

dataset = pd.read_csv("D:/Sx Varun/Sx Code/MCA/2nd_Tri/python/dataSet/iris.csv")
dataset1 = dataset

print(dataset1.head(10))

dataset.columns = dataset.columns.str.strip()
```

```python
plt.plot(dataset.index, dataset['SepalLength'], color='blue', label='SepalLength')
plt.plot(dataset.index, dataset['SepalWidth'], color='orange', label='SepalWidth')
plt.title('Sepal Length and Sepal Width vs Index')
plt.xlabel('Index')
plt.ylabel('Value')
plt.legend()
plt.grid()
plt.show()
```

```python
plt.scatter(dataset['SepalLength'], dataset['SepalWidth'], color='red', label='SepalWidth')
plt.scatter(dataset['SepalLength'], dataset['PetalLength'], color='green', label='PetalLength')
plt.scatter(dataset['SepalLength'], dataset['PetalWidth'], color='blue', label='PetalWidth')

# Chart formatting
plt.title('Iris Dataset Features vs SepalLength')
plt.xlabel('SepalLength')
plt.ylabel('Value')
plt.grid()
plt.legend()
plt.show()
```

```python
mean_sepal_length = dataset.groupby('Species')['SepalLength'].mean()

colors = ['skyblue', 'lightgreen', 'salmon']

mean_sepal_length.plot(kind='bar', color=colors)

# Chart formatting
plt.title('Mean Sepal Length by Species')
plt.xlabel('Species')
plt.ylabel('Mean Sepal Length')
plt.grid(axis='y', linestyle='--', alpha=0.7)

# for index, value in enumerate(mean_sepal_length):
    # plt.text(index, value + 0.02, round(value, 2), ha='center', fontsize=9)

plt.show()
```

```python
#Histogram
plt.hist(dataset['PetalLength'], bins=20, color='purple', edgecolor='black', alpha=0.7)
plt.title('Distribution of Petal Length')
plt.xlabel('Petal Length')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
#Pie Chart
species_counts = dataset['Species'].value_counts()
colors = ['gold', 'lightcoral', 'lightskyblue']
species_counts.plot(kind='pie', autopct='%1.1f%%', colors=colors, startangle=140, shadow=True)
plt.title('Species Distribution in Iris Dataset')
plt.ylabel('') 
plt.axis('equal')  
plt.show()
```

```python
# Group by species and calculate average SepalLength
avg_sepal_length = dataset.groupby('Species')['SepalLength'].mean()

# Pie chart values
values = avg_sepal_length.values  
labels = avg_sepal_length.index  

# Colors
colors = ['gold', 'lightcoral', 'lightskyblue']

# Plot pie chart
plt.figure(figsize=(6,6))
plt.pie(values,
        labels=labels,
        autopct='%1.1f%%',
        colors=colors,
        startangle=140,
        shadow=True)

plt.title('Average SepalLength per Species')
plt.axis('equal')
plt.show()
```

```python
#Box Plot
plt.figure(figsize=(8, 6))
dataset.boxplot(column='PetalWidth', by='Species', grid=False, patch_artist=True,
                boxprops=dict(facecolor='lightblue', color='blue'),
                medianprops=dict(color='red'),
                whiskerprops=dict(color='blue'),
                capprops=dict(color='blue'),
                flierprops=dict(marker='o', color='blue', alpha=0.5))
plt.title('Petal Width Distribution by Species')
plt.suptitle('')
plt.xlabel('Species')
plt.ylabel('Petal Width (cm)')
plt.legend(['Petal Width'])
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pandas import melt  # optional (pd.melt works too)

# Set Seaborn theme
sns.set_theme(style="whitegrid")

# Create subplots
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# ──────────────────────────────
# 1️⃣ Line Plot: Sepal Length & Width vs Index
# ──────────────────────────────
sns.lineplot(x=dataset.index, y=dataset['SepalLength'], color='blue', label='SepalLength', ax=axes[0, 0])
sns.lineplot(x=dataset.index, y=dataset['SepalWidth'], color='orange', label='SepalWidth', ax=axes[0, 0])
axes[0, 0].set_title('Sepal Length and Sepal Width vs Index')
axes[0, 0].legend()

# ──────────────────────────────
# 2️⃣ Scatter Plot: SepalLength vs other features
# ──────────────────────────────
sns.scatterplot(x='SepalLength', y='SepalWidth', data=dataset, color='red', label='SepalWidth', ax=axes[0, 1])
sns.scatterplot(x='SepalLength', y='PetalLength', data=dataset, color='green', label='PetalLength', ax=axes[0, 1])
sns.scatterplot(x='SepalLength', y='PetalWidth', data=dataset, color='blue', label='PetalWidth', ax=axes[0, 1])
axes[0, 1].set_title('Iris Dataset Features vs SepalLength')
axes[0, 1].legend()

# ──────────────────────────────
# 3️⃣ Bar Plot: Mean Sepal Length by Species
# ──────────────────────────────
mean_sepal_length = dataset.groupby('Species')['SepalLength'].mean().reset_index()
sns.barplot(
    x='Species', y='SepalLength', hue='Species', data=mean_sepal_length,
    palette=['skyblue', 'lightgreen', 'salmon'], ax=axes[1, 0], legend=False
)
axes[1, 0].set_title('Mean Sepal Length by Species')
axes[1, 0].set_xlabel('Species')
axes[1, 0].set_ylabel('Mean Sepal Length')

# ──────────────────────────────
# 4️⃣ Box Plot: All Features
# ──────────────────────────────
# Melt dataset for Seaborn-friendly format
melted = pd.melt(dataset,
                 id_vars='Species',
                 value_vars=['SepalLength', 'SepalWidth', 'PetalLength', 'PetalWidth'],
                 var_name='Feature',
                 value_name='Value')

sns.boxplot(
    x='Feature', y='Value', hue='Feature', data=melted,
    ax=axes[1, 1], palette='Set2', legend=False
)
axes[1, 1].set_title('Box Plot of Iris Features')

# Layout
plt.tight_layout()
plt.show()
```

```python
# Histogram (Distribution Plot)

plt.figure(figsize=(8, 5))
sns.histplot(dataset['PetalLength'], bins=20, kde=True, color='purple')
plt.title('Distribution of Petal Length')
plt.xlabel('Petal Length')
plt.ylabel('Frequency')
plt.show()
```

```python
# KDE Plot (Kernel Density Estimate)

plt.figure(figsize=(8, 5))
sns.kdeplot(dataset['SepalLength'], shade=True, color='green')
sns.kdeplot(dataset['PetalLength'], shade=True, color='orange')
plt.title('KDE Plot of Sepal & Petal Length')
plt.xlabel('Length')
plt.show()
```

```python
# Pair Plot (Scatter Matrix)

sns.pairplot(dataset, hue='Species', palette='husl')
plt.suptitle('Pair Plot of Iris Dataset', y=1.02)
plt.show()
```

```python
# Heatmap (Correlation Matrix)

plt.figure(figsize=(8, 6))
sns.heatmap(dataset.drop(columns='Species').corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap of Iris Dataset')
plt.show()
```

```python
# Violin Plot

plt.figure(figsize=(8, 6))
sns.violinplot(x='Species', y='PetalLength', data=dataset, palette='pastel')
plt.title('Violin Plot of Petal Length by Species')
plt.show()
```

```python
# Swarm Plot / Strip Plot

plt.figure(figsize=(8, 6))
sns.swarmplot(x='Species', y='SepalWidth', data=dataset, palette='muted')
plt.title('Swarm Plot of Sepal Width by Species')
plt.show()
```

```python
# Joint Plot

sns.jointplot(x='SepalLength', y='PetalLength', data=dataset, kind='scatter', color='teal')
plt.suptitle('Joint Plot: SepalLength vs PetalLength', y=1.02)
plt.show()
```
