```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y1 = [2, 4, 1, 8, 7]
y2 = [5, 3, 4, 2, 1]
y3 = [3, 5, 7, 6, 4]
y4 = [8, 6, 4, 2, 1]

plt.figure(figsize=(8, 6))

# 1st subplot – line chart
plt.subplot(2, 2, 1)
plt.plot(x, y1)
plt.title('Line Chart')

# 2nd subplot – scatter
plt.subplot(2, 2, 2)
plt.scatter(x, y2)
plt.title('Scatter Plot')

# 3rd subplot – bar
plt.subplot(2, 2, 3)
plt.bar(x, y3)
plt.title('Bar Chart')

# 4th subplot – histogram
plt.subplot(2, 2, 4)
plt.hist(y4)
plt.title('Histogram')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Example DataFrame
df = pd.DataFrame({
    'A': [1, 2, 3, 4, 5],
    'B': [2, 3, 4, 5, 6],
    'C': [5, 4, 3, 2, 1]
})

cols = df.columns
n = len(cols)

fig, axes = plt.subplots(n, n, figsize=(8, 8))

for i in range(n):
    for j in range(n):
        if i == j:
            # Diagonal – Histogram of the column
            axes[i, j].hist(df[cols[i]], bins=5)
            axes[i, j].set_ylabel('')
            axes[i, j].set_xlabel(cols[i])
        else:
            # Off-diagonal – Scatter Plot
            axes[i, j].scatter(df[cols[j]], df[cols[i]])
            axes[i, j].set_xlabel('')
            axes[i, j].set_ylabel('')
        
        # Remove tick labels for cleaner look
        axes[i, j].tick_params(labelbottom=False, labelleft=False)

plt.tight_layout()
plt.show()
```
