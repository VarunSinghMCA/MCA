```python
import pandas as pd
```

```python
Employee_info = {
    'EmployeeID': [1, 2, 3, 4],
    'Name': ['Ram', 'Ravi', 'Alen', 'Divya'],
    'Department': ['HR', 'Engineering', 'HR', 'IT'],
}

df_emp_info = pd.DataFrame(Employee_info)

print(df_emp_info)
```

```python
Salary_info = {
    'EmployeeID': [1,2,4,5],
    'Salary': [60000, 80000, 95000, 70000]
}

df_salary_info = pd.DataFrame(Salary_info)

print(df_salary_info)
```

```python
New_Employee_info = {
    'Name': ['Neelam', 'Fredic'],
    'Department': ['IT', 'HR'],
}

df_new_emp_info = pd.DataFrame(New_Employee_info)

print(df_new_emp_info)
```

Merges: Inner, Outer, Left and Right Joins

```python
# By default Merge joins on 'inner'
default_merged_df = pd.merge(df_emp_info, df_salary_info, on='EmployeeID')
print(default_merged_df)
```

```python
outer_merged_df = pd.merge(df_emp_info, df_salary_info, on='EmployeeID', how='outer')

print(outer_merged_df)
```

```python
inner_merged_df = pd.merge(df_emp_info, df_salary_info, on='EmployeeID', how='inner')

print(inner_merged_df)
```

```python
right_merged_df = pd.merge(df_emp_info, df_salary_info, on='EmployeeID', how='right')
print(right_merged_df)
```

```python
left_merged_df = pd.merge(df_emp_info, df_salary_info, on='EmployeeID', how='left')
print(left_merged_df)
```

```python
# Merge employee info with New Employee info on EmployeeID
merged_new_emp_df = pd.merge(df_emp_info, df_new_emp_info, on='Department', how='outer', indicator=True)
print(merged_new_emp_df)
```

```python
# First, perform a full outer join between df_emp_info and df_salary_info
outer_full_join_df = pd.merge(df_emp_info, df_salary_info, on='EmployeeID', how='outer', indicator=True)
print(outer_full_join_df)

# Then, concatenate the new employees (who do not have EmployeeID or Salary info)
concat_full_df = pd.concat([outer_full_join_df[['Name', 'Department', 'Salary']], df_new_emp_info], ignore_index=True)
print(concat_full_df)
```

JOINS

```python
# Inner Join
joined_df = df_emp_info.set_index('EmployeeID').join(df_salary_info.set_index('EmployeeID'), how='inner')
print(joined_df)
```

```python
# Outer Join
outer_joined_df = df_emp_info.set_index('EmployeeID').join(df_salary_info.set_index('EmployeeID'), how='outer')
print(outer_joined_df)
```

```python
# Left Join
left_joined_df = df_emp_info.set_index('EmployeeID').join(df_salary_info.set_index('EmployeeID'), how='left')
print(left_joined_df)
```

```python
# Right Join
right_joined_df = df_emp_info.set_index('EmployeeID').join(df_salary_info.set_index('EmployeeID'), how='right')
print(right_joined_df)
```

Concatenation

```python
# Concatenation
concat_df = pd.concat([df_emp_info, df_new_emp_info], ignore_index=True)
print(concat_df)
```

```python
# Concatination with ignore index
concat_ignore_index_df = pd.concat([df_emp_info, df_new_emp_info], ignore_index=True)
print(concat_ignore_index_df)
```

```python
# Concat 2 df side-by-side
# Assume df has similar indexes of allignment
concat_side_by_side_df = pd.concat([df_emp_info, df_salary_info], axis=1)
print(concat_side_by_side_df)
```
