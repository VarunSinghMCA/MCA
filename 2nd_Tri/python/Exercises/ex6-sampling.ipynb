```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
import seaborn as sns

data = pd.read_csv("D:/Sx Varun/Sx Code/MCA/2nd_Tri/python/dataSet/employee_compensation.csv")
dataset = data

print(dataset.head(10))

dataset.columns = dataset.columns.str.strip()
```

Random Sampling

```python
sample_simple = dataset.sample(n=10, random_state=1)
print("\n",sample_simple)
```

Systematic Sampling

```python
def systematic_sampling(df, step):
    indices = np.arange(0, len(df), step)
    return df.iloc[indices]

def sys_sampling(df, step):
    start = np.random.randint(0, step)
    indices = np.arange(start, len(df), step)
    return df.iloc[indices]

systematic_sample = systematic_sampling(dataset, step=10)
print("\n",systematic_sample)
```

Stratified Sampling

```python
# Take a unique value from a categorical column to create strata
strata = dataset['DEPARTMENT_NAME'].unique()
# Remove random state for true randomness in stratified sampling
sample_stratified = dataset.groupby('DEPARTMENT_NAME', group_keys=False).apply(lambda x: x.sample(2, random_state=1))
print("\n",sample_stratified)
```

Cluster Samplling

```python
# Randomly Select 3 clusters
clusters = dataset['DEPARTMENT_NAME'].unique()
selected_clusters = np.random.choice(clusters, size=3, replace=False)
cluster_sample = dataset[dataset['DEPARTMENT_NAME'].isin(selected_clusters)]
print("\n",cluster_sample)
```

```python
choose_cluster = np.random.choice(dataset['DEPARTMENT_NAME'].unique(), 2, replace=False)
clustered_data = dataset[dataset['DEPARTMENT_NAME'].isin(choose_cluster)]
print("\n",clustered_data)
print("\n",clustered_data['DEPARTMENT_NAME'].unique())
```
