```python
# Load data (adjust path)
missdata1 = pd.read_csv("D:/Sx Varun/Sx Code/MCA/2nd_Tri/python/dataSet/num1.csv")

# Strip any trailing spaces in column names
missdata1.columns = missdata1.columns.str.strip()

# Basic info
print(missdata1.describe())                     # statistical summary
print(missdata1.info())                         # info about datatypes and non-null values
print(missdata1.columns)                        # list of columns
print(missdata1.dtypes)                         # data types of each column
print(missdata1.shape)                          # rows and columns
print(missdata1.size)                           # total number of elements
print(missdata1.isnull().sum())                 # count of missing values in each column

# Missing values
print(missdata1.dropna())                       # drop rows with missing values
print(missdata1.fillna(0))                      # fill missing values with 0

# Statistics on MALE column
print(missdata1['MALE'].mean())                 # mean of column MALE
print(missdata1['MALE'].median())               # median of column MALE
print(missdata1['MALE'].mode())                 # mode of column MALE

# Sorting / grouping
print(missdata1.sort_values(by='MALE'))         # sort by column MALE
print(missdata1.groupby('AGE').mean(numeric_only=True))  # group by AGE and calculate mean

# Corrected unique
print(missdata1['MALE'].unique())               # unique values in column MALE
print(missdata1['MALE'].value_counts())         # count of unique values in column MALE

# Sampling
print(missdata1.sample(5))                      # random sample of 5 rows

# Correlation / covariance
print(missdata1.corr(numeric_only=True))        # correlation matrix
print(missdata1.cov(numeric_only=True))         # covariance matrix

# Duplicates
print(missdata1.duplicated().sum())             # count of duplicate rows
print(missdata1.drop_duplicates())              # drop duplicate rows

# Apply functions
print(missdata1.apply(np.sqrt, numeric_only=True))  # apply square root to numeric columns
print(missdata1.applymap(lambda x: x*2 if isinstance(x, (int, float)) else x))  # double each number
print(missdata1['MALE'].map(lambda x: x*2))     # apply function to single column

# Renaming / index changes
print(missdata1.rename(columns={'MALE':'Alpha', 'B':'Beta'}))
print(missdata1.set_index('MALE'))             # set column MALE as index
print(missdata1.reset_index())                 # reset index to default

# Pivot and melt (make sure columns exist)
if 'B' in missdata1.columns and 'C' in missdata1.columns:
    print(missdata1.pivot_table(values='C', index='MALE', columns='B', aggfunc='mean'))
    print(missdata1.melt(id_vars=['MALE'], value_vars=['B', 'C']))

# Save to CSV
missdata1.to_csv('output.csv', index=False)

# Head / tail / indexing
print(missdata1.head())
print(missdata1.tail())
print(missdata1.sample(5))
print(missdata1.iloc[0:5, 0:2])                # first 5 rows and first 2 columns
print(missdata1.loc[missdata1['MALE'] > 50])   # rows where MALE > 50
print(missdata1.at[0, 'MALE'])                 # value at row 0, column MALE
print(missdata1.iat[0, 0])                     # value at row 0, column 0

# Conditions
print(missdata1.where(missdata1 > 50))         # values >50, others NaN
print(missdata1.mask(missdata1 > 50))          # values >50 replaced with NaN
print(missdata1.sample(frac=0.1))              # random 10% sample of rows

# Memory usage
print(missdata1.memory_usage(deep=True))       # memory usage of dataframe
```
